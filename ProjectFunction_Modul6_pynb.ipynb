```python
uang = 10000
item = ['nibati','agua','worchestershire','Stands','Poki','cookie','worse']
harga = [3000, 3000, 5500, 3500, 2000, 4000, 6000]
# stock= [3, 4, 5, 1, 1, 2, 4]
pesanan = []

print('uang jajan : ',uang)
def menu():
  for i in range(len(item)):
    print(i+1,'[',item[i],']','=',harga[i])
    i += 1

def PILIH():
  global pilih
  while True:
    pilih = int(input('Masukkan pilihan anda (0 untuk selesai): '))

    if pilih == 0:
          break

    elif 1 <= pilih <= len(item):
        pesanan.append(item[pilih - 1])
        print("Anda memilih:", item[pilih - 1])
    else:
        print("Pilihan tidak valid, coba lagi.")

    print("\nPesanan Anda:")
    for i in pesanan:
        print("-", i)


def total():
    global total_harga
    total_harga = 0
    for p in pesanan:
        hargabarang = item.index(p)
        total_harga += harga[hargabarang]
    print("\nTotal harga =", total_harga)

def bayar():
    global uang, total_harga, kembalian
    kembalian = uang - total_harga
    if kembalian < 0:
        print("\n⚠️ Uang Anda Tidak Cukup! Pesanan dibatalkan.")
        pesanan.clear()
        menu()
        PILIH()
    else:
      print("\nUang Anda =", kembalian)


menu()
PILIH()
total()
bayar()





```
