```python
import matplotlib.pyplot as Plot

Buah = ['Apel', 'Pisang','Durian', 'Salak','Nanas','Semangka', 'Alpukat']
Jumlah = [30,25,50,17,32,30,16]
Plot.figure(figsize = (10,7))
Plot.plot(Buah, Jumlah, marker='x', color='red', label='Stok Buah Sehat Bergizi dan Manis')
Plot.title('Data Stok Buah')
Plot.xlabel('Buah')
Plot.ylabel('Jumlah')
Plot.legend()
Plot.grid(True)

Plot.figure(figsize=(10,7))
Plot.bar(Buah, Jumlah, color='green')
Plot.title('Stok Buah Sehat Bergizi dan Manis')
Plot.xlabel('Buah')
Plot.ylabel('Jumlah')
Plot.show()
```
