```python
print ('PROGRAM PEMONITORAN KEADAAN PADA RUANG MESIN')
suhu =[] #List masukkan sensor suhu
magnitudo = [] # List masukkan sensor gempa
Warning = 10 #Berapa kali "Warning Message" perlu diulangi
while True:
    main = int(input('======[MAIN CONSOLE]======\n1. MAGNITUDE MONITOR\n2. TEMPERATURE MONITOR\n3. VIEW SUMMARY\n4. EXIT\n'))
    if main == 3: #Menunjukkan rangkuman dari semua data yang didapatkan kedua sensor
      if len(magnitudo) == 0:
        print('DATA MAGNITUDO KOSONG')
      else: #Menentukan nilai rata-rata, tertinggi dan, terendah sensor gempa
        magnitudo_rtrt = sum(magnitudo)/len(magnitudo)
        Magnitudo = max(magnitudo)
        magnitudO = min(magnitudo)
        print('-------------------------') #Menunjukan rangkuman nilai sensor
        print('MAGNITUDO RATA-RATA :',magnitudo_rtrt)
        print('MAGNITUDO TERBESAR :',Magnitudo)
        print('MAGNITUDO TERKECIL :',magnitudO)
        print('-------------------------')
        if Magnitudo >= 4 and Magnitudo <= 4.9: #Kondisi yang dianggap "tidak normal"
          print('Terdeteksi Gempa Sedang')
        if Magnitudo >= 5 and Magnitudo <= 5.9:
          print('Terdeteksi Gempa Siaga')
        if Magnitudo >= 6 :
          print('Terdeteksi Gempa Besar')
          for i in range(Warning):
            print('SEEK SHELTER IMMIDIEATELY')
            print('THIS IS NOT A DRILL')
          break # Magnitudo 4 sampai 6 ke atas dianggap gempa yang lumayan was-was

      if len(suhu) == 0:
        print('DATA SUHU KOSONG\n')
      else: #Menentukan nilai rata-rata, tertinggi dan, terendah sensor suhu
        suhu_rtrt = sum(suhu)/len(suhu)
        Suhu = max(suhu)
        suhU = min(suhu)
        print('-------------------------') #Menunjukan rangkuman nilai sensor
        print('SUHU RATA-RATA :',suhu_rtrt)
        print('SUHU TERBESAR :',Suhu)
        print('SUHU TERKECIL :',suhU)
        print('-------------------------')
      if suhu_rtrt >= 35: #Kondisi yang dianggap "tidak normal"
        print('Temperatur Panas')
        print('Tolong turunkan pengaturan suhu AC')
      if suhu_rtrt <= 17:
          print('It is quite cold')
          print('Turn on  the Heater')
      if suhu_rtrt >= 45:
          print('IT IS HOT IT IS HOT')
          for i in range(Warning):
            print('HIGH TEMPERATURE DETECTED')
            print('PLEASE TURN DOWN THE AC')
            print('BUKA JENDELA BUKA JENDELA')
      if suhu_rtrt <= 10:
          print('IT IS COLD IT IS COLD')
          for i in range(Warning):
            print('ABNORMALLY LOW TEMPERATURE DETECTED')
            print('PLEASE TURN UP THE HEATER')
            print('START A FIRE, SEEK WARMTH IMMIDIATELY')
      continue # Beda dengan warning gempa, warning suhu tidak memberhentikan program

    if main == 4: #Keluar dari program
        print('LOG OUT')
        break


#--------------MENU GEMPA---------------#
    while True:
      if main == 1: #Kondisi ini menunjukan menu monitor gempa
        print('\n')
        GEMPA = int(input('======[MAGNITUDE MONITOR]======\n1. Add Data\n2. Edit Data\n3. Remove Data\n4. EXIT\n'))
        if GEMPA == 1: # Kondisi untuk memasukkan data untuk sensor gempa
            while True :
                print('\n')
                print('Masukkan nilai "0" (nol) untuk kembali ke menu') #Masukkan nilai 0 (nol) kapanpun untuk "cancel"
                nambah_M = float(input('Masukkan nilai magnitudo saat ini :'))
                print('Masukkan nilai "0" (nol) untuk kembali ke menu')
                if nambah_M > 0 or nambah_M < 0:
                    magnitudo.append(nambah_M)
                    print('---------------------') #Menunjukkan data gempa
                    for i in magnitudo:
                        print(i);
                    print('---------------------')
                else :
                    break

        if GEMPA == 2: #Kondisi untuk mengubah (edit) data gempa
            while True :
              print('---------------------')
              for i in magnitudo:
                  print(i);
              print('---------------------')
              ubah_M = int(input('Masukkan nilai keberapa yang ingin diubah : Nilai ke-'))
              if ubah_M == 0:
                break

              else:
                ubah_M -= 1 # Dikurang 1 karena list dihitung dari nol
                nilai_ubah_M = float(input('Diubah menjadi berapa :'))
                magnitudo[ubah_M] = nilai_ubah_M #Menggantikan nilai lama dengan yang baru
                print('---------------------')
                for i in magnitudo:
                    print(i);
                print('---------------------')
                print('Masukkan nilai "0" (nol) untuk kembali ke menu')
                if nilai_ubah_M == 0:
                    break

        if GEMPA == 3: # Kondisi untuk menghapus salah satu bacaan sensor
            while True :
              print('---------------------')
              for i in magnitudo:
                  print(i);
              print('---------------------')
              ngurang_M = int(input('Masukkan nilai keberapa yang ingin dihapus :'))
              if ngurang_M == 0:
                  break
              ngurang_M -= 1
              magnitudo.pop(ngurang_M) #Mengurangi nilai dari urutan sesuai input
              print('---------------------')
              for i in magnitudo:
                  print(i);
              print('---------------------')

        if GEMPA == 4: # Kembali ke menu [MAIN CONSOLE]
          print ('list akhir :\n')
          print('---------------------')
          for i in magnitudo:
              print(i);
          print('---------------------')
          break

  #------------MENU SUHU---------------#
      if main == 2:
        print('\n')
        SUHU = int(input('======[TEMPERATURE MONITOR]======\n1. Add Data\n2. Edit Data\n3. Remove Data\n4. EXIT\n'))
        if SUHU == 1:
            while True :
                print('\n')
                print('Masukkan nilai "0" (nol) untuk kembali ke menu')
                nambah_S = float(input('Masukkan nilai suhu saat ini :'))
                print('Masukkan nilai "0" (nol) untuk kembali ke menu')
                if nambah_S > 0 or nambah_S < 0:
                    suhu.append(nambah_S)
                    print('---------------------')
                    for i in suhu:
                        print(i);
                    print('---------------------')
                else :
                    break

        if SUHU == 2:
            while True :
              print('---------------------')
              for i in suhu:
                  print(i);
              print('---------------------')
              ubah_S = int(input('Masukkan nilai keberapa yang ingin diubah : Nilai ke-'))
              if ubah_S == 0:
                break

              else:
                ubah_S -= 1
                nilai_ubah_S = float(input('Diubah menjadi berapa :'))
                suhu[ubah_S] = nilai_ubah_S
                print('---------------------')
                for i in suhu:
                    print(i);
                print('---------------------')
                print('Masukkan nilai "0" (nol) untuk kembali ke menu')
                if nilai_ubah_S == 0:
                    break

        if SUHU == 3:
            while True :
              print('---------------------')
              for i in suhu:
                  print(i);
              print('---------------------')
              ngurang_S = int(input('Masukkan nilai keberapa yang ingin dihapus :'))
              if ngurang_S == 0:
                  break
              ngurang_S -= 1
              suhu.pop(ngurang_S)
              print('---------------------')
              for i in suhu:
                  print(i);
              print('---------------------')

        if SUHU == 4:
          print ('list akhir :\n')
          print('---------------------')
          for i in suhu:
              print(i);
          print('---------------------')
          break
#---------------------------------------------------------------------------------------#


```
